### ***========================Grading System========================*** 

```python

print("===== STUDENT GRADING SYSTEM =====")

name = input("Enter your Full_Name: ").title()

marks = int(input("Enter your marks (0-100): "))

if marks >= 90 and marks <= 100:
    print("Student Name:", name)
    print("Grade: A1")
    print("Remarks: Outstanding Performance")

elif marks >= 80:
    print("Student Name:", name)
    print("Grade: A+")
    print("Remarks: Excellent Performance")

elif marks >= 70:
    print("Student Name:", name)
    print("Grade: A")
    print("Remarks: Good Performance")

elif marks >= 60:
    print("Student Name:", name)
    print("Grade: B")
    print("Remarks: Satisfactory Performance")

elif marks >= 50:
    print("Student Name:", name)
    print("Grade: C")
    print("Remarks: Normal Performance")

elif marks >= 40:
    print("Student Name:", name)
    print("Grade: D")
    print("Remarks: Keep It Up")

elif marks >= 33 and marks < 40:
    print("Student Name:", name)
    print("Grade: E")
    print("Remarks: Need Improvement")

elif marks >= 0 and marks <= 32:
    print("Student Name:", name)
    print("Failed:")
    print("Remarks: Work Hard")

else:
    print("Invalid Marks! Please enter marks between 0 and 100.")
```

### Write a program that asks the user to enter a number.

If the number is even and greater than 50, print "Even and Big".

If the number is even and less than or equal to 50, print "Even and Small".

If the number is odd and greater than 50, print "Odd and Big".

Otherwise, print "Odd and Small".

```python
try:
    Num = int(input('Enter a number: '))

    if Num % 2 == 0 and Num > 50:
        print('Even and Big')

    elif Num % 2 == 0 and Num <= 50:
        print('Even and Small')

    elif Num % 2 != 0 and Num > 50:
        print('Odd and Big')

    else:
        print('Odd and Small')

except ValueError:
    print('Invalid input! Please enter a valid number.')
```

```python

print('=========== E-Commerce Shopping & Discount System ===========')

customer_name = input('Enter customer name: ').title()

total_Amt = float(input('Enter your shopping amount: '))

memberShip_status = input('Enter membership (Gold/Silver/Regular): ').lower()

coupon = input('Do you have a coupon? (Yes/No): ').lower()


# Step 1: Shopping amount discount

if total_Amt >= 100000:
    discount = 20

elif total_Amt >= 50000:
    discount = 15

elif total_Amt >= 20000:
    discount = 10

else:
    discount = 0


# Step 2: Membership discount

if memberShip_status == 'gold':
    discount += 5

elif memberShip_status == 'silver':
    discount += 3

else:
    discount += 0


# Step 3: Coupon discount

if coupon == 'yes':
    discount += 2


# Step 4: Calculate final bill

discount_amount = total_Amt * discount / 100

final_amount = total_Amt - discount_amount


# Step 5: Print final bill

print('\n========== CUSTOMER FINAL BILL ==========')

print('Customer Name       :', customer_name)
print('Original Amount     : Rs.', format(total_Amt, ',.2f'))
print('Total Discount      :', discount, '%')
print('Discount Amount     : Rs.', format(discount_amount, ',.2f'))
print('Final Payable Amount: Rs.', format(final_amount, ',.2f'))

print('=========================================')
print('Thank you for shopping!')
```

```python

print("===== BANK LOAN APPROVAL SYSTEM =====")

Customer_Name = input('Enter your Customer name: ').title()

Customer_Age = int(input('Enter your age: '))
Monthly_salary = int(input('Enter your Monthly salary: '))
experience = int(input('Enter your working experience: '))
Credit_score = int(input('Enter your Credit score: '))

# Checking eligibility

if Customer_Age >= 21 and Customer_Age <= 50:

    if Monthly_salary >= 50000:

        if experience >= 2:

            if Credit_score >= 750:

                print(f'\n{Customer_Name}, You are eligible for Premium Loan')

            elif Credit_score >= 650:

                print(f'\n{Customer_Name}, You are eligible for Normal Loan')

            else:
                print('Loan Rejected: Credit score is insufficient')

        else:
            print('Experience insufficient')

    else:
        print('Salary insufficient')

else:
    print('You are not eligible because of your age')
```
